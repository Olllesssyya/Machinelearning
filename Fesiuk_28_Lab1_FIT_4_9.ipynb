1. Створіть одновимірний масив (вектор) зі значеннями n! кожного натурального числа n, починаючи з 1, та виведіть його значення.

```python
count = 10
n_values = np.arange(1, count + 1)
factorial_array = np.cumprod(n_values)
print(f"Вектор: {factorial_array}")
```

2. Створіть двовимірний масив (матрицю) розміром 3x3, заповніть його оберненими значеннями косинусів кожного натурального числа, починаючи з 1, та виведіть його значення.

```python
n_values = np.arange(1, 10)
inv_cos = 1 / np.cos(n_values)
matrix = inv_cos.reshape((3, 3))
print(f"Матриця 3x3 з оберненими значеннями косинусів від 1 до 9: \n {matrix}")
```

3. Створіть масив розміром 3x3, заповніть його випадковими дійсними числами від -1 до 1 та знайдіть відсоток чисел, які менше нуля.

```python
matrix = np.random.uniform(-1, 1, (3, 3))
neg_count = np.sum(matrix < 0)
total = matrix.size
percent_neg = (neg_count / total) * 100

print("Згенерована матриця:")
print(matrix)
print(f"Відсоток чисел менших за нуль: {percent_neg:.2f}%")
```

5. Створіть два одновимірних масиви розміром 5, заповніть їх випадковими цілими числами в діапазоні від 2 до 7 та виконайте на них поелементні операції додавання, віднімання та множення.

```python
array1 = np.random.randint(2, 8, size=5)
array2 = np.random.randint(2, 8, size=5)

sum_result = array1 + array2
diff_result = array1 - array2
prod_result = array1 * array2

print("Масив 1:", array1)
print("Масив 2:", array2)
print("Сума:", sum_result)
print("Різниця:", diff_result)
print("Добуток:", prod_result)
```

14.Створіть матрицю розміром 4x4, заповніть її випадковими цілими числами в діапазоні від 1 до 100 та знайдіть суму елементів на побічній діагоналі.

```python
matrix = np.random.randint(1, 101, (4, 4))
sum = np.trace(np.fliplr(matrix))

print("Matrix:\n", matrix)
print("Сума елементів на побічній діагоналі:", sum)
```

15.Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні від -2 до 2 та знайдіть вектор-стовпчик, що містить найменше за модулем значення елементу у кожному стовпці матриці.

```python
matrix = np.random.uniform(-2, 2, (3, 3))
idx = np.argmin(np.abs(matrix), axis=0)
vector = matrix[idx, np.arange(matrix.shape[1])].reshape(-1, 1)

print("Matrix:\n", matrix)
print("Вектор-стовпчик найменших за модулем елементів кожного стовпця:\n", vector)
```
